```python
def classic_bubble_sort(my_list):
    n = len(my_list)
    comparisons = 0

    for i in range(n - 1):
        for j in range(n - 1 - i):
            comparisons += 1
            if my_list[j] > my_list[j + 1]:                  
                my_list[j], my_list[j + 1] = my_list[j + 1], my_list[j]  

    return my_list, comparisons
```

```python
# Lists for tests
arr1 = [1, 2, -3, 4, -4, 5, 50]
arr2 = [8, 91, -1000, 141, 22, 67, -67, 59, 625, 228]
```

```python
sorted_arr1, comp_arr1 = classic_bubble_sort(arr1)
sorted_arr2, comp_arr2 = classic_bubble_sort(arr2)

print(f'First array sorted: {sorted_arr1}), first array number of comparisons: {comp_arr1}')
print(f'Second array sorted: {sorted_arr2}), second array number of comparisons: {comp_arr2}')
```
