```python
def classic_bubble_sort(my_list):
    n = len(my_list)
    comparisons = 0

    for i in range(n - 1):
        for j in range(n - 1 - i):
            comparisons += 1
            if my_list[j] > my_list[j + 1]:                  
                my_list[j], my_list[j + 1] = my_list[j + 1], my_list[j]  

    return my_list, comparisons
```

```python
# Lists for tests
short_list = [1, 2, -3, 4, -4, 5, 50]
long_list = [8, 91, -1000, 141, 22, 67, -67, 59, 625, 228]
```

```python
sorted_short_list, comp_short_list = classic_bubble_sort(short_list)
sorted_long_list, comp_long_list = classic_bubble_sort(long_list)

print(f'First array sorted: {sorted_short_list}), first array number of comparisons: {comp_short_list}')
print(f'Second array sorted: {sorted_long_list}), second array number of comparisons: {comp_long_list}')
```
